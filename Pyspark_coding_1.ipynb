#####Data Reading

```python
from pyspark.sql.functions import *
```

```python
df = spark.sql("""select *from workspace.bigdata.big_mart_sales""")
```

```python
df.select('Item_Identifier','Item_Weight','Item_Fat_Content').show(3) #Just selecetd few columns
```

```python
df.select(col('Item_Identifier'),col('Item_Weight')).show(3)
```

```python
df.select('*').show(2)
```

#####Alias tranformation

```python
df.select(col('Item_Identifier').alias("Item_ID")).show(2) 
```

#####Filter tranformation

#### Scenario 1

```python

df.filter(col('Item_Fat_Content')=='Regular').display()
```

####Scenario2

```python
df.filter((col('Item_Type')=='Soft Drinks') & (col('Item_Weight')<10)).show(3)
```

######Secnario3

```python
df.filter(col('Outlet_Size').isNull() & col('Outlet_Location_Type').isin('Tier 1','Tier 2')).show(5)
```

#####WithcolumnRename this changes columns names at the dataframe level and you can use same name for future transformation

```python
df.withColumnRenamed('Item_Weight','Item_Wt').show(3)
```

#####WithColumn If you want to create new column you can use with Column function lit() is used to keep the contant values

#######Scenario1

```python
df.withColumn('Flag',lit('New')).show(5)
```

```python
df.withColumn('Multiply',col('Item_Weight')*col('Item_MRP')).show(5)
```

#####Scenario2

```python
df.withColumn('Item_Fat_Content',regexp_replace(col('Item_Fat_Content'),'Low Fat','LF'))\
    .withColumn('Item_Fat_Content',regexp_replace(col('Item_Fat_Content'),'Regular','Reg')).show(5)
```

```python
df=df.withColumn('Item_weight',col('Item_Weight').cast(StringType()))
```

```python
df.printSchema()
```

##Sorting

####Scenario 2

#####Scenario--3 Sorting based on 2 columns descending

#####Scenario-- 4

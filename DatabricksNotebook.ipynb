## DWH 

```sql
create database salesDWH;
```

### Staging Layer

```sql
--Initial load
create table salesdwh.stg_sales 
as 
select * from sales.orders;
```

### Transformation

```sql
create or replace table salesdwh.stg_sales
as 
select * from sales.orders where orderdate>'2024-02-10';
```

```sql
select * from salesdwh.stg_sales;
```

```sql
select * from salesdwh.stg_sales;
```

####Transformations

```sql
create view salesdwh.trans_sales
as 
select * from salesdwh.stg_sales where Quantity is not Null;
```

```sql
select * from salesdwh.trans_sales;
```

### core layer

```sql
create table salesdwh.core_sales
as
select * from salesdwh.trans_sales;
```

```sql
select * from salesdwh.core_sales;
```

#### Display DWH core layer

```sql
select * from salesdwh.core_sales;
```
